```python
# AI-Assisted Threat Detection Dashboard
# SQL Tasks using SQLite (Google Colab)

import sqlite3
import pandas as pd

# TASK 1: Create Database

conn = sqlite3.connect("ThreatDetectionDB.db")
cursor = conn.cursor()

print("Database Created Successfully!")

# TASK 2: Create SecurityEvents Table

cursor.execute("""

CREATE TABLE IF NOT EXISTS SecurityEvents (

Event_ID INTEGER PRIMARY KEY,
Timestamp TEXT,
User_Name TEXT,
IP_Address TEXT,
Region TEXT,
Event_Type TEXT,
Severity TEXT,
Status TEXT

)

""")

print("Table Created Successfully!")


```

```python
# TASK 3: Insert 10 Sample Security Events

sample_data = [

(1,'2026-07-01 09:00','Alice','192.168.1.10','Asia','Login Failure','Low','Resolved'),

(2,'2026-07-01 09:15','Bob','192.168.1.20','Europe','Malware','High','Open'),

(3,'2026-07-01 09:30','Charlie','192.168.1.30','Asia','Phishing','Medium','Resolved'),

(4,'2026-07-01 09:45','David','192.168.1.20','Europe','Unauthorized Access','Critical','Open'),

(5,'2026-07-01 10:00','Emma','192.168.1.40','North America','Login Failure','Low','Resolved'),

(6,'2026-07-01 10:15','Frank','192.168.1.50','Asia','DDoS','High','Investigating'),

(7,'2026-07-01 10:30','Grace','192.168.1.20','Europe','Malware','Critical','Open'),

(8,'2026-07-01 10:45','Henry','192.168.1.60','Africa','Phishing','Medium','Resolved'),

(9,'2026-07-01 11:00','Isabella','192.168.1.70','North America','Data Leak','High','Investigating'),

(10,'2026-07-01 11:15','Jack','192.168.1.80','Asia','Login Failure','Low','Resolved')

]

cursor.executemany("""

INSERT OR REPLACE INTO SecurityEvents
VALUES (?,?,?,?,?,?,?,?)

""", sample_data)

conn.commit()

print("10 Security Events Inserted Successfully!")

# TASK 4: Display All Security Events

print("\n========== ALL SECURITY EVENTS ==========\n")

df = pd.read_sql_query("SELECT * FROM SecurityEvents", conn)

print(df)
```

```python

# TASK 5: Threat Severity Analysis

print("\n========== THREAT SEVERITY ANALYSIS ==========\n")

query = """

SELECT Severity,
COUNT(*) AS Total_Threats

FROM SecurityEvents

GROUP BY Severity

ORDER BY Total_Threats DESC

"""

print(pd.read_sql_query(query, conn))

# TASK 6: Event Type Analysis

print("\n========== EVENT TYPE ANALYSIS ==========\n")

query = """

SELECT Event_Type,
COUNT(*) AS Total_Events

FROM SecurityEvents

GROUP BY Event_Type

ORDER BY Total_Events DESC

"""

print(pd.read_sql_query(query, conn))
```

```python
# TASK 7: Repeat Attackers Analysis

print("\n========== REPEAT ATTACKERS ==========\n")

query = """

SELECT
IP_Address,
COUNT(*) AS Number_of_Attacks

FROM SecurityEvents

GROUP BY IP_Address

HAVING COUNT(*) > 1

"""

print(pd.read_sql_query(query, conn))

# TASK 8: Region-Based Threat Analysis

print("\n========== REGION-BASED THREAT ANALYSIS ==========\n")

query = """

SELECT
Region,
COUNT(*) AS Total_Threats

FROM SecurityEvents

GROUP BY Region

ORDER BY Total_Threats DESC

"""

print(pd.read_sql_query(query, conn))

# TASK 9: Open Security Incidents

print("\n========== OPEN SECURITY INCIDENTS ==========\n")

query = """

SELECT *

FROM SecurityEvents

WHERE Status='Open'

"""

print(pd.read_sql_query(query, conn))

```

```python
# TASK 10: Dashboard Summary


print("\n========== DASHBOARD SUMMARY ==========\n")

query = """

SELECT

COUNT(*) AS Total_Events,

SUM(CASE WHEN Severity='Critical' THEN 1 ELSE 0 END) AS Critical_Threats,

SUM(CASE WHEN Severity='High' THEN 1 ELSE 0 END) AS High_Threats,

SUM(CASE WHEN Severity='Medium' THEN 1 ELSE 0 END) AS Medium_Threats,

SUM(CASE WHEN Severity='Low' THEN 1 ELSE 0 END) AS Low_Threats,

SUM(CASE WHEN Status='Open' THEN 1 ELSE 0 END) AS Open_Incidents,

SUM(CASE WHEN Status='Resolved' THEN 1 ELSE 0 END) AS Resolved_Incidents,

SUM(CASE WHEN Status='Investigating' THEN 1 ELSE 0 END) AS Investigating_Incidents

FROM SecurityEvents

"""

print(pd.read_sql_query(query, conn))

# Close Database

conn.close()

print("\nDatabase Closed Successfully!")
print("\nAll SQL Tasks Executed Successfully!")
```
