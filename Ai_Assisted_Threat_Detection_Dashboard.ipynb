**Data Analysis**

```python

# Step 5: Count Events by Severity

print("\nNumber of Events by Severity")
severity_count = df["Severity"].value_counts()
print(severity_count)

# Step 6: Count Events by Event Type

print("\nNumber of Events by Event Type")
event_count = df["Event_Type"].value_counts()
print(event_count)

# Step 7: Display Only High Severity Events

print("\nHigh Severity Events")
high_events = df[df["Severity"] == "High"]
print(high_events)

```
